```python
lista_negra = ["basura", "malo", "tonto", "asco", "maldito", "hediondo"]

def evaluar_comentario(texto, palabras_prohibidas, clasificador_ia, negativo=0.50):
    # 1. Comprobar lista negra
    texto_limpio = texto.lower()
    palabras_detectadas = [p for p in palabras_prohibidas if p in texto_limpio]

    if palabras_detectadas:
        return False, f"RECHAZADO: Contiene palabra(s) prohibida(s): {palabras_detectadas}"

    # 2. Obtener predicción del modelo
    prediccion = clasificador_ia(texto)[0]
    etiqueta = prediccion['label']
    score = prediccion['score']

    # 3. Aplicar regla para comentarios negativos
    if etiqueta == 'NEG' and score > negativo:
        return False, f"RECHAZADO: Comentario negativo detectado ({score:.1%} de certeza)"

    # 4. Aceptar positivos y neutros sin palabras prohibidas
    return True, f"ACEPTADO: Sentimiento {etiqueta} ({score:.1%} de certeza)"

# Probar la lógica con todos los casos
ejemplos = [
    "Me encanta este módulo de IA",
    "Este servicio es una completa basura, estoy muy decepcionado",
    "El paquete llegó hoy a las 3 de la tarde",
    "Qué genial, se arruinó el coche justo cuando iba a salir de vacaciones",
    "Eres un hediondo me llegó roto el pedido 👎"
]

print("\n=== RESULTADOS DEL FILTRADO ===")
for t in ejemplos:
    aceptado, motivo = evaluar_comentario(t, lista_negra, clasificador)
    print(f"\nTexto: \"{t}\"\n-> {motivo}")
```
